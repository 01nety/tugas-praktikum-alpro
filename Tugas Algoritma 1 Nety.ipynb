# Praktikum:Toko Dorayaki Doraemon
##### Toko dorayaki doraemon adalah toko yang di bangun oleh Doraemon dengan dibantu oleh ke-3 teman lainnya, yaitu Nobita, Suneo, dan Shizuka. program ini akan menjabarkan bagaimana Doraemon membangun tokonya, mulai dari penamaan, perhitungan modal, hingga perhitungan keuntungan dalam sebulan.

```python
nama_makanan = "ayam "
nama_karakter = "doraemon "
kata_sifat = "jujur "
nama_toko = nama_makanan + nama_karakter + kata_sifat
print(nama_toko) 
#untuk membuat variabel dengan spasi harus ditambahkan (_), selanjutnya tambahkan (=) untuk menunjukan value dari variabel, lalu tambahkan (") yang mengapit value, terakhir gunakan tanda (+) untuk menyatukan masing-masing value.
```

```python
modal_doraemon = 555000
modal_nobita = 333000
modal_shizuka = 222000
total_modal = modal_doraemon + modal_nobita + modal_shizuka
print("total modal", total_modal)
#untuk menambahkan value berupa angka tidak perlu menggunakan ("), lalu menghitung modal gunakan (+), terakhir untuk menampilkan pemrograman gunakan print.
```

```python
jumlah_dorayaki = 350
harga_dorayaki = 5000
total_pendapatan = jumlah_dorayaki * harga_dorayaki
print("total pendapatan", total_pendapatan)
#untuk menghitung total pendapatan selama sebulan masing-masing variabel akan dioperasikan dengan menggunakan tanda (*)
```

```python
keuntungan = total_pendapatan - total_modal
print("keuntungan", keuntungan)
#untuk menghitung keuntungan menggunakan rumus mencari keuntungan, untuk tanda yang digunakan (-)
```

```python
print("keuntungan total adalah", keuntungan)
#kesalahan nobita ada pada kekurangannya (,) setelah kalimat sehingga program tidak berhasil dijalankan.
```
