```python
skor_nobita = 0

def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai
    
# 1. Fungsi Belajar
def belajar(jam_belajar):
    if jam_belajar < 1:
        print("Nobita hanya membuka buku sebentar")
        return 50
    elif 1 <= jam_belajar < 3:
        print("Nobita belajar secukupnya")
        return 25
    else:
        print("Nobita belajar dengan rajin hari ini")
        return 10

# 2. Fungsi Tidur
def tidur(jam_tidur):
    if jam_tidur < 1:
        print("Nobita kurang tidur")
        return -10
    elif 1 <= jam_tidur <= 2:
        print("Nobita tidur terlalu lama")
        return 20
    else:
        print("Nobita tidur terlalu lama")
        return -20

# 3. Fungsi main bersama
def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print("Nobita Bermain secukupnya")
        return 15
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print("Nobita masih bisa mengatur waktu dengan baik")
        return 5
    else:
        print("Nobita kebanyakan bermain")
        return -30

# 4. Fungsi menyontek
def menyontek():
    print("Nobita ketahuan menyontek")
    return -50

# Fungsi pengendali Aktivitas
def jalankan_aktivitas(pilihan):
    if pilihan == 1:
        jam_belajar = float(input("Berapa jam nobita belajar hari ini? "))
        nilai = belajar(jam_belajar)

    elif pilihan == 2:
        jam_tidur = float(input("Berapa jam nobita tidur siang hari ini? "))
        nilai =tidur(jam_tidur)

    elif pilihan == 3:
        jumlah_gadget = int(input("Berapa gadget yang digunakan nobita? "))
        waktu_main = int(input("Berapa jam nobita bermain gadget? ")) 
        nilai = main_bersama_doraemon(jumlah_gadget, waktu_main)

    elif pilihan == 4:
        nilai = menyontek()

    elif pilihan > 4:
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")
        return

    update_skor(nilai)

# Fungsi evaluasi akhir
def evaluasi_akhir(skor):
    if skor >= 80:
        return "siswa teladan"
    elif 50 <= skor < 80:
        return "cukup rajin"
    else:
        return "pemalas"

# Menu Utama
print("=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")
print("1. Belajar")
print("2. Tidur")
print("3. Main bersama doraemon")
print("4. menyontek saat ujian")
print()

pilihan = int(input("Pilihan aktivitas (1-4): "))

jalankan_aktivitas(pilihan)

print("\n=== HASIL EVALUASI ===")
print("Skor Nobita:", skor_nobita)
print("Evaluasi:", evaluasi_akhir(skor_nobita))
```
