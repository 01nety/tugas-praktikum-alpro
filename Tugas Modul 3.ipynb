```python
belanjaan = ["Apel", "Susu", "Roti"]
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)
print(f" list belanjaan: {belanjaan}")
print(f" harga jumlah: {harga_jumlah}")

karakter_pertama = belanjaan[0][0]
karakter_terakhir = belanjaan[-1][-1]
print(f" karakter pertama dari barang pertama ({belanjaan[0]}): {karakter_pertama}")
print(f" karakter terakhir dari barang terakhir ({belanjaan[-1]}): {karakter_terakhir}")
```

```python
total_apel = harga_jumlah[0] * harga_jumlah[1]
total_susu = harga_jumlah[2] * harga_jumlah[3]
total_roti = harga_jumlah[4] * harga_jumlah[5]
total_keseluruhan = total_apel + total_susu + total_roti

print(str(f" total apel ({belanjaan[0]}): {harga_jumlah[0]} * {harga_jumlah[1]} = {total_apel}"))
print(str(f" total susu ({belanjaan[1]}): {harga_jumlah[2]} * {harga_jumlah[3]} = {total_susu}"))
print(str(f" total roti ({belanjaan[2]}): {harga_jumlah[4]} * {harga_jumlah[5]} = : {total_roti}"))
print(str(f" total keseluruhan: {total_keseluruhan}"))
```

```python
nama_pembeli = "Nobita_Nobi"
nama_toko = "Toko_Sembako_Doraemon"

print("=======================================")
print(f"\t{nama_toko}")
print("=======================================")
print(f"Pembeli : {nama_pembeli}")
print("---------------------------------------")
print("Barang\tHarga\tJumlah\tTotal")
print("---------------------------------------")
print(f"{belanjaan[0]}\t{harga_jumlah[0]}\t{harga_jumlah[1]}\t{total_apel}")
print(f"{belanjaan[1]}\t{harga_jumlah[2]}\t{harga_jumlah[3]}\t{total_susu}")
print(f"{belanjaan[2]}\t{harga_jumlah[4]}\t{harga_jumlah[5]}\t{total_roti}")
print("---------------------------------------")
print(f"Total")
```
